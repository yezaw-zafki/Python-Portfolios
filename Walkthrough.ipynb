```python
import pandas as pd

def clean_data(df):
    # Drop duplicate rows across all columns
    df = df.drop_duplicates()
    # Convert column names to lowercase
    df.columns = df.columns.str.lower()
    # Rename column 'passengerid' to 'passenger_id'
    df = df.rename(columns={'passengerid': 'passenger_id'})
    # Split name into first and last
    df[['last_name', 'first_name']] = df['name'].str.split(',', expand=True)
    # Drop column: 'name'
    df = df.drop(columns=['name'])
    # columns relocation
    df = df[['passenger_id', 'survived', 'pclass', 'first_name', 'last_name', 'sex', 'age', 'sibsp', 'parch', 'ticket', 'fare', 'cabin', 'embarked']]
    # Replace missing age values by sex mean
    mean_age = df.groupby('sex')['age'].mean().round()
    df['age'] = df['age'].fillna(df['sex'].map(mean_age))
    # Replace missing values with "N/A" in column: 'cabin'
    df = df.fillna({'cabin': "N/A"})
    # Replace missing values with the most common value of each column in: 'embarked'
    df = df.fillna({'embarked': df['embarked'].mode()[0]})
    return df

# Loaded variable 'df' from URI: /Users/yezawmyint/.vscode/extensions/ms-toolsai.datawrangler-1.24.2/walkthrough.csv
df = pd.read_csv(r'/Users/yezawmyint/.vscode/extensions/ms-toolsai.datawrangler-1.24.2/walkthrough.csv')

df_clean = clean_data(df.copy())
df_clean.head()
```
